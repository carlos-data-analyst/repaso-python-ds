Tarea 2: Lógica de Negocio (Bucle + Condicional)

    Crea una lista llamada precios con al menos 5 valores numéricos.
    Itera sobre la lista y, usando un condicional, imprime si cada precio es "Caro" (mayor a 100) o "Económico" (menor o igual a 100).


```python
precios = [50,75,100,125,150]

for i in precios:
    if i > 100:
        print(f"El precio de {i} es caro.")
    else:
        print(f"El precio de {i} es económico")
```

Tarea 4: Modularización con Funciones

    Escribe una función llamada resumen_estadistico que reciba una lista de números.
    La función debe devolver un diccionario que contenga: la suma total, el promedio y la cantidad de elementos.
    Llama a la función con una lista de ejemplo e imprime el resultado.


```python
def resumen_estadistico(*numeros):
    sumatoria = sum(numeros)
    promedio = sum(numeros)/len(numeros)
    cantidad = len(numeros)
    
    diccionario = ({
        "sumatoria": sumatoria,
        "promedio": promedio,
        "cantidad": cantidad
    })
    
    return diccionario

resumen_estadistico(100,200,300)
```
